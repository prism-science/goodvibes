```python
#In this notebook, I want to construct the stiffness matrix for a helical spring and compute the effective spring that describes its motion.
#I will follow Steve's method to create the force constant matrix for rigid bodies from the one acting on point particles.
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from mpl_toolkits.mplot3d.art3d import Poly3DCollection
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
from scipy.linalg import expm

#Construct matrix r_x such that r_x v = r x v.
def cross_matrix(r):
    return np.array([[    0, -r[2],  r[1]],
                     [ r[2],     0, -r[0]],
                     [-r[1],  r[0],     0]])

#Create matrix that changes point mass force constants to rigid body force constants.
def A(r):
    return np.block([np.eye(3), -cross_matrix(r)])

#Create force constant matrix for simple springs connecting point masses.
def Phi(r1, r2, kL, kT):
    n = (r1-r2)/np.linalg.norm(r1-r2)
    phi = kL*np.outer(n,n)+kT*(np.eye(3)-np.outer(n,n))
    return phi

#Add up the right on and off-diagonal blocks that a spring contributes.
def pair_hessian(xA, xB, rA, rB, kL, kT):
    """
    xA, xB : equilibrium positions of the two contact points in world coordinates
    rA, rB : vectors from the respective rigid-body reference points
             to the contact points
    """

    phi = Phi(xA, xB, kL, kT)

    AA = A(rA)
    AB = A(rB)

    HAA = AA.T @ phi @ AA
    HAB = -AA.T @ phi @ AB
    HBA = -AB.T @ phi @ AA
    HBB = AB.T @ phi @ AB

    return np.block([
        [HAA, HAB],
        [HBA, HBB]
    ])

#Construct matrix B that takes one from 12D infinitesimal motion coordinates to 6D relative pose coordinates.
#
#The reference point at which the relative pose (u_rel, theta_rel) is expressed is the
#midpoint between the two rigid-body centers, C = (RA+RB)/2 = RA + d/2 = RB - d/2.
#This is the "democratic" choice: neither body is privileged, and for a bond with an
#axis of symmetry along d it coincides with the true center of stiffness (the point
#where the translation-rotation coupling block of K is symmetric). It is derived by
#transporting each body's rigid motion to the midpoint:
#   u_rel = [uB + thB x (C-RB)] - [uA + thA x (C-RA)]
#         = -uA + (X/2) thA + uB + (X/2) thB
#   th_rel = thB - thA   (reference-point independent, as it must be)
def relative_pose_matrix(d):
    I = np.eye(3)
    Z = np.zeros((3, 3))
    X = cross_matrix(d)

    return np.block([[-I,  X/2,  I,  X/2],
                     [ Z,  -I,  Z,   I]])

#General version: reference point P = midpoint + r, for locating/verifying the true
#center of stiffness (the r that zeroes the antisymmetric part of K_tr) as opposed to
#just the geometric midpoint (r=0). RA, RB are the two body centers; d = RB - RA.
def relative_pose_matrix_general(RA, RB, r=np.zeros(3)):
    I = np.eye(3)
    Z = np.zeros((3, 3))
    C = (RA + RB) / 2 + r
    XA = cross_matrix(C - RA)
    XB = cross_matrix(C - RB)

    return np.block([[-I,  XA,  I, -XB],
                     [ Z,  -I,  Z,   I]])


def relative_stiffness(H, d):
    B = relative_pose_matrix(d)
    Bplus = np.linalg.pinv(B)

    Krel = Bplus.T @ H @ Bplus

    return Krel

#Geometry of the problem: two unit cubes centered at (0,0,0) and (0,0,2).
#Connect their vertices in such a way to make helical spring.
#examine the structure of the 6x6 matrix to understand what's going on.

#centers of cubes
RA = np.array([0., 0., 0.])
RB = np.array([0., 0., 2.])

#positions of connectors relative to cube centers
a1 = np.array([ 0.25,  0.25, 0.5])
a2 = np.array([-0.25,  0.25, 0.5])
a3 = np.array([-0.25, -0.25, 0.5])
a4 = np.array([ 0.25, -0.25, 0.5])

b1 = np.array([ 0.25,  0.25, -0.5])
b2 = np.array([-0.25,  0.25, -0.5])
b3 = np.array([-0.25, -0.25, -0.5])
b4 = np.array([ 0.25, -0.25, -0.5])

#lab frame positions of spring connections
A_points = [RA + a for a in [a1, a2, a3, a4]]
B_points = [RB + b for b in [b1, b2, b3, b4]]

#make left- and right-handed helical springs: make 12x12 matrix of derivatives.
kL = 1.0
#adding kT violates SE(3) invariance.
kT = 0.0

Hr = pair_hessian(A_points[0], B_points[1], a1, b2, kL, kT) + pair_hessian(A_points[1], B_points[2], a2, b3, kL, kT) + pair_hessian(A_points[2], B_points[3], a3, b4, kL, kT) + pair_hessian(A_points[3], B_points[0], a4, b1, kL, kT) + pair_hessian(A_points[0], B_points[0], a1, b1, kL, kT) + pair_hessian(A_points[1], B_points[1], a2, b2, kL, kT) + pair_hessian(A_points[2], B_points[2], a3, b3, kL, kT) + pair_hessian(A_points[3], B_points[3], a4, b4, kL, kT)
Hl = pair_hessian(A_points[0], B_points[3], a1, b4, kL, kT) + pair_hessian(A_points[1], B_points[0], a2, b1, kL, kT) + pair_hessian(A_points[2], B_points[1], a3, b2, kL, kT) + pair_hessian(A_points[3], B_points[2], a4, b3, kL, kT) + pair_hessian(A_points[0], B_points[0], a1, b1, kL, kT) + pair_hessian(A_points[1], B_points[1], a2, b2, kL, kT) + pair_hessian(A_points[2], B_points[2], a3, b3, kL, kT) + pair_hessian(A_points[3], B_points[3], a4, b4, kL, kT)
```

```python
B = relative_pose_matrix(np.array([0., 0., 2.]))

Krel_r = relative_stiffness(Hr, np.array([0., 0., 2.]))
Krel_l = relative_stiffness(Hl, np.array([0., 0., 2.]))

Hr_reconstructed = B.T @ Krel_r @ B
Hl_reconstructed = B.T @ Krel_l @ B

#The Hessians are reconstructable as long as kT=0.
errorr = np.linalg.norm(Hr - Hr_reconstructed)
print(errorr)

errorl = np.linalg.norm(Hl - Hl_reconstructed)
print(errorl)
```

```python
#======================================================================
# Alternative: same animation, but bonds rendered as solid pipes
#======================================================================
# Reuses the translucent-cube / depth-sorting machinery above, but replaces
# each flat bond line with a shaded cylindrical tube (built the same way as
# the chiral-pipe visualization for K, minus the helical stripe -- a single
# point-to-point spring has no internal torsional structure of its own, so
# there is nothing to twist; the "pipe" here is just a solid rod standing in
# for what was a 2D line, so it participates properly in the depth sort).

def _cylinder_faces(p0, p1, radius, n_theta=10):
    """Lateral quad faces of a cylindrical pipe from p0 to p1 (no end caps)."""
    axis = p1 - p0
    L = np.linalg.norm(axis)
    if L < 1e-9:
        return []
    ahat = axis / L
    arbitrary = np.array([0.,0.,1.]) if abs(ahat[2]) < 0.9 else np.array([1.,0.,0.])
    nvec = arbitrary - np.dot(arbitrary, ahat)*ahat
    nvec /= np.linalg.norm(nvec)
    bvec = np.cross(ahat, nvec)
    theta = np.linspace(0, 2*np.pi, n_theta, endpoint=True)
    ring0 = p0 + radius*(np.outer(np.cos(theta), nvec) + np.outer(np.sin(theta), bvec))
    ring1 = p1 + radius*(np.outer(np.cos(theta), nvec) + np.outer(np.sin(theta), bvec))
    return [np.array([ring0[i], ring0[i+1], ring1[i+1], ring1[i]]) for i in range(n_theta-1)]


def _draw_scene_pipes(ax, VA, VB, VAb, VBb, bonds1, bonds2,
                      cubeA_color='tab:blue', cubeB_color='tab:orange',
                      cube_alpha=0.35, pipe_radius=0.045):
    """Same depth-sorted painter's algorithm as _draw_scene, but bonds are
    solid shaded tubes (Poly3DCollection) instead of Line3D segments."""
    vhat = _camera_direction(ax)
    drawables = []

    for V, color in [(VA, cubeA_color), (VB, cubeB_color)]:
        for idx in _CUBE_FACES:
            verts = V[list(idx)]
            depth = np.dot(verts.mean(axis=0), vhat)
            drawables.append((depth, 'face', (verts, color)))

    for (i, j) in bonds1:
        p0, p1 = VAb[i], VBb[j]
        faces = _cylinder_faces(p0, p1, pipe_radius)
        drawables.append((np.dot(0.5*(p0+p1), vhat), 'tube', (faces, 'firebrick')))
    for (i, j) in bonds2:
        p0, p1 = VAb[i], VBb[j]
        faces = _cylinder_faces(p0, p1, pipe_radius)
        drawables.append((np.dot(0.5*(p0+p1), vhat), 'tube', (faces, 'darkorange')))

    drawables.sort(key=lambda d: d[0])

    for depth, kind, payload in drawables:
        if kind == 'face':
            verts, color = payload
            poly = Poly3DCollection([verts], facecolor=color, edgecolor='k',
                                     linewidths=0.6, alpha=cube_alpha)
            ax.add_collection3d(poly)
        else:
            faces, color = payload
            if not faces:
                continue
            tube = Poly3DCollection(faces, facecolors=[color]*len(faces),
                                     edgecolor='none', alpha=1.0, shade=True)
            ax.add_collection3d(tube)


def animate_relative_coordinate_pipes(d, coordinate,
                                      amplitude=0.25,
                                      nframes=60,
                                      interval=40,
                                      cube_alpha=0.35,
                                      pipe_radius=0.045):
    """Identical to animate_relative_coordinate, except the bonds are drawn
    as solid pipes (via _draw_scene_pipes) rather than flat lines."""

    B = relative_pose_matrix(d)
    Bplus = np.linalg.pinv(B)
    e = np.zeros(6); e[coordinate] = 1.0
    q = Bplus @ e
    uA, thA, uB, thB = q[0:3], q[3:6], q[6:9], q[9:12]

    xA = np.array([0., 0., 0.])
    xB = np.array(d, dtype=float)

    vertices = np.array([
        [ 0.5,  0.5,  0.5], [-0.5,  0.5,  0.5], [-0.5, -0.5,  0.5], [ 0.5, -0.5,  0.5],
        [ 0.5,  0.5, -0.5], [-0.5,  0.5, -0.5], [-0.5, -0.5, -0.5], [ 0.5, -0.5, -0.5]
    ])
    bondvertices = np.array([
        [ 0.25,  0.25,  0.5], [-0.25,  0.25,  0.5], [-0.25, -0.25,  0.5], [ 0.25, -0.25,  0.5],
        [ 0.25,  0.25, -0.5], [-0.25,  0.25, -0.5], [-0.25, -0.25, -0.5], [ 0.25, -0.25, -0.5]
    ])
    bonds1 = [(0, 5), (1, 6), (2, 7), (3, 4)]
    bonds2 = [(0, 4), (1, 5), (2, 6), (3, 7)]

    fig = plt.figure(figsize=(6, 6))
    ax = fig.add_subplot(111, projection="3d", computed_zorder=False)

    names = [
        "relative x translation", "relative y translation", "relative z translation",
        "relative x rotation", "relative y rotation", "relative z rotation",
    ]

    def update(frame):
        s = amplitude * np.sin(2*np.pi * frame / nframes)

        RA = expm(s * cross_matrix(thA))
        RB = expm(s * cross_matrix(thB))

        VA = xA + s*uA + (RA @ vertices.T).T
        VB = xB + s*uB + (RB @ vertices.T).T
        VAb = xA + s*uA + (RA @ bondvertices.T).T
        VBb = xB + s*uB + (RB @ bondvertices.T).T

        ax.clear()
        _draw_scene_pipes(ax, VA, VB, VAb, VBb, bonds1, bonds2,
                          cube_alpha=cube_alpha, pipe_radius=pipe_radius)

        centerA = xA + s*uA
        centerB = xB + s*uB
        ax.scatter(*centerA, s=40, c='tab:blue', depthshade=False)
        ax.scatter(*centerB, s=40, c='tab:orange', depthshade=False)

        ax.set_xlim(-1.5, 1.5)
        ax.set_ylim(-1.5, 1.5)
        zmin = min(-1.0, d[2] - 1.0)
        zmax = max( 1.0, d[2] + 1.0)
        ax.set_zlim(zmin, zmax)
        ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
        ax.set_title(names[coordinate])
        ax.set_box_aspect([3, 3, max(3, zmax-zmin)])

    ani = FuncAnimation(fig, update, frames=nframes, interval=interval, blit=False)
    plt.close(fig)
    return HTML(ani.to_jshtml())
```

```python
#======================================================================
# Center-of-stiffness solver (needed for the single-pipe animation below)
#======================================================================
def vec_from_antisym(M):
    """Inverse of cross_matrix: extract w such that cross_matrix(w) == 0.5*(M - M.T)."""
    return 0.5*np.array([M[2,1]-M[1,2], M[0,2]-M[2,0], M[1,0]-M[0,1]])

def shift_K(K, r):
    """Exact transform of a bond stiffness K to a new reference point offset by r."""
    Ktt = K[:3,:3]; Ktr = K[:3,3:]; Krr = K[3:,3:]
    X = cross_matrix(r)
    Ktr_new = Ktt @ X + Ktr
    Krr_new = Krr - X@Ktt@X - X@Ktr + Ktr.T@X
    Knew = np.zeros((6,6))
    Knew[:3,:3] = Ktt; Knew[:3,3:] = Ktr_new; Knew[3:,:3] = Ktr_new.T; Knew[3:,3:] = Krr_new
    return Knew

def find_center_of_stiffness(K, rcond=1e-10):
    """Returns (K_cos, r, info) -- see the earlier center-of-stiffness cells for full docs."""
    Ktt = K[:3,:3]; Ktr = K[:3,3:]
    M = np.zeros((3,3))
    for k in range(3):
        ek = np.zeros(3); ek[k] = 1.0
        M[:,k] = vec_from_antisym(Ktt @ cross_matrix(ek))
    w_target = -vec_from_antisym(Ktr)
    r, *_ = np.linalg.lstsq(M, w_target, rcond=rcond)
    K_cos = shift_K(K, r)
    Ktr_cos = K_cos[:3,3:]
    chirality = 0.5*(Ktr_cos + Ktr_cos.T)
    info = {'residual_norm': np.linalg.norm(0.5*(Ktr_cos - Ktr_cos.T)),
            'chirality': chirality, 'chirality_norm': np.linalg.norm(chirality)}
    return K_cos, r, info

def _rotation_minimizing_frame(points):
    """Parallel-transported (approximately rotation-minimizing) Frenet-like frame
    along a discretized curve -- avoids the spurious extra twisting a naive
    per-point Frenet frame introduces."""
    tangents = np.zeros_like(points)
    tangents[1:-1] = points[2:] - points[:-2]
    tangents[0] = points[1] - points[0]
    tangents[-1] = points[-1] - points[-2]
    tangents /= np.linalg.norm(tangents, axis=1, keepdims=True)

    t0 = tangents[0]
    arbitrary = np.array([0.,0.,1.]) if abs(t0[2]) < 0.9 else np.array([1.,0.,0.])
    n0 = arbitrary - np.dot(arbitrary,t0)*t0
    n0 /= np.linalg.norm(n0)
    normals = np.zeros_like(points); normals[0] = n0

    for i in range(1, len(points)):
        tp, tc = tangents[i-1], tangents[i]
        nprev = normals[i-1]
        v = np.cross(tp,tc); c = np.clip(np.dot(tp,tc), -1, 1)
        if np.linalg.norm(v) < 1e-12:
            nnew = nprev
        else:
            vhat = v/np.linalg.norm(v); angle = np.arctan2(np.linalg.norm(v), c)
            nnew = (nprev*np.cos(angle) + np.cross(vhat,nprev)*np.sin(angle)
                    + vhat*np.dot(vhat,nprev)*(1-np.cos(angle)))
        nnew = nnew - np.dot(nnew,tc)*tc
        nnew /= np.linalg.norm(nnew)
        normals[i] = nnew

    binormals = np.cross(tangents, normals)
    return tangents, normals, binormals
```

```python
#======================================================================
# Animation using a SINGLE aggregate helical pipe (built from K), instead of
# the individual point-spring connectors
#======================================================================
# This computes the 6x6 relative stiffness K once (e.g. Krel_r), finds its
# center of stiffness and intrinsic chirality, and represents the ENTIRE bond
# network between the two cubes as one pipe: bowed through the center of
# stiffness, with a helical stripe whose handedness/pitch come from K's
# chirality. As the animation moves the two cubes, K is re-expressed in the
# current orientation by rotating it with the MEAN rotation of the two bodies
# (block_diag(R_mean, R_mean)^T @ K @ block_diag(R_mean, R_mean)) -- the same
# "treat both bodies democratically" convention already used for the
# relative-pose frame itself. This is an approximation when the two bodies
# rotate by different amounts (there is no single rigid rotation that exactly
# carries the whole bond network along), but it is the natural, consistent
# choice given that convention, and is exact whenever the two bodies rotate
# together or the motion is a pure translation.

def _block_diag_rot(R):
    Z = np.zeros((3,3))
    return np.block([[R, Z],[Z, R]])

def _helix_pipe_drawables(K, RA, RB, vhat, n_pts=45, n_theta=10, radius=None,
                          radius_ref=None, n_turns_scale=6.0, pipe_color=None):
    RA = np.asarray(RA,float); RB = np.asarray(RB,float)
    d = RB-RA; L = np.linalg.norm(d); dhat = d/L

    K_cos, r_star, info = find_center_of_stiffness(K)
    C_mid = (RA+RB)/2; C_star = C_mid + r_star
    chi = info['chirality']

    P1 = 2*C_star - (RA+RB)/2
    t = np.linspace(0,1,n_pts)[:,None]
    points = (1-t)**2*RA + 2*t*(1-t)*P1 + t**2*RB
    tangents, normals, binormals = _rotation_minimizing_frame(points)

    Ktt = K[:3,:3]; Krr = K[3:,3:]
    if radius is None:
        scale = np.trace(Ktt) if radius_ref is None else np.trace(Ktt)/radius_ref
        radius = 0.06*L*max(scale,1e-6)**0.25
    if pipe_color is None:
        ax_stiff = dhat@Ktt@dhat
        perp_stiff = (np.trace(Ktt)-ax_stiff)/2
        aniso = np.log((ax_stiff+1e-9)/(perp_stiff+1e-9))
        pipe_color = plt.cm.coolwarm(0.5+0.15*np.clip(aniso,-3,3))

    theta = np.linspace(0,2*np.pi,n_theta,endpoint=True)
    rings = []
    for i in range(n_pts):
        circ = points[i][:,None] + radius*(np.outer(normals[i],np.cos(theta))
                                            + np.outer(binormals[i],np.sin(theta)))
        rings.append(circ.T)

    drawables = []
    for i in range(n_pts-1):
        r0, r1 = rings[i], rings[i+1]
        for j in range(n_theta-1):
            face = np.array([r0[j], r0[j+1], r1[j+1], r1[j]])
            depth = np.dot(face.mean(axis=0), vhat)
            drawables.append((depth, 'pipeface', (face, pipe_color)))

    evals, evecs = np.linalg.eigh(chi)
    order = np.argsort(-np.abs(evals)); evals, evecs = evals[order], evecs[:,order]
    axial_idx = np.argmax(np.abs(evecs.T @ dhat))
    lam_par = evals[axial_idx]
    perp_idx = [i for i in range(3) if i != axial_idx]
    lam_perp = np.mean(evals[perp_idx])

    Krr_par = dhat@Krr@dhat
    Krr_perp = (np.trace(Krr)-Krr_par)/2
    arclen = np.concatenate([[0], np.cumsum(np.linalg.norm(np.diff(points,axis=0),axis=1))])
    s_arc = arclen/arclen[-1]

    def stripe_points(lam, Krr_ref):
        if abs(lam) < 1e-8:
            return None
        pitch = np.sign(lam)*np.sqrt(abs(lam)/max(Krr_ref,1e-9))
        n_turns = n_turns_scale*np.tanh(abs(pitch))
        angle = 2*np.pi*n_turns*s_arc*np.sign(pitch)
        return points + radius*1.03*(normals*np.cos(angle)[:,None] + binormals*np.sin(angle)[:,None])

    for lam, Krr_ref, color, lw in [(lam_par, Krr_par, 'crimson', 3.0),
                                     (lam_perp, Krr_perp, 'darkorange', 1.8)]:
        sp = stripe_points(lam, Krr_ref)
        if sp is None:
            continue
        for i in range(len(sp)-1):
            p0, p1 = sp[i], sp[i+1]
            depth = np.dot(0.5*(p0+p1), vhat)
            drawables.append((depth, 'line', (p0, p1, color, lw)))

    meta = {'r_star': r_star, 'C_star': C_star, 'chirality': chi,
            'lam_parallel': lam_par, 'lam_perp': lam_perp}
    return drawables, meta


def _draw_scene_helix(ax, VA, VB, K_t, RA_t, RB_t, cubeA_color='tab:blue',
                      cubeB_color='tab:orange', cube_alpha=0.35, **pipe_kwargs):
    vhat = _camera_direction(ax)
    drawables = []
    for V, color in [(VA, cubeA_color), (VB, cubeB_color)]:
        for idx in _CUBE_FACES:
            verts = V[list(idx)]
            depth = np.dot(verts.mean(axis=0), vhat)
            drawables.append((depth, 'cubeface', (verts, color)))

    pipe_drawables, meta = _helix_pipe_drawables(K_t, RA_t, RB_t, vhat, **pipe_kwargs)
    drawables.extend(pipe_drawables)
    drawables.sort(key=lambda x: x[0])

    for depth, kind, payload in drawables:
        if kind == 'cubeface':
            verts, color = payload
            ax.add_collection3d(Poly3DCollection([verts], facecolor=color, edgecolor='k',
                                                   linewidths=0.6, alpha=cube_alpha))
        elif kind == 'pipeface':
            face, color = payload
            ax.add_collection3d(Poly3DCollection([face], facecolors=[color],
                                                   edgecolor='none', alpha=0.95, shade=True))
        else:
            p0, p1, color, lw = payload
            ax.plot3D([p0[0],p1[0]], [p0[1],p1[1]], [p0[2],p1[2]], color=color, linewidth=lw)
    return meta


def animate_relative_coordinate_helix(d, coordinate, K,
                                      amplitude=0.25, nframes=36, interval=60,
                                      cube_alpha=0.35, **pipe_kwargs):
    """
    Same relative-coordinate animation as before, but the bond network is
    represented by a single aggregate helical pipe built from its 6x6
    relative stiffness K (e.g. Krel_r, Krel_b, Krel_o), rather than by its
    individual point-spring connectors.
    """
    B = relative_pose_matrix(d)
    Bplus = np.linalg.pinv(B)
    e = np.zeros(6); e[coordinate] = 1.0
    q = Bplus @ e
    uA, thA, uB, thB = q[0:3], q[3:6], q[6:9], q[9:12]

    xA = np.array([0., 0., 0.])
    xB = np.array(d, dtype=float)

    vertices = np.array([
        [ 0.5,  0.5,  0.5], [-0.5,  0.5,  0.5], [-0.5, -0.5,  0.5], [ 0.5, -0.5,  0.5],
        [ 0.5,  0.5, -0.5], [-0.5,  0.5, -0.5], [-0.5, -0.5, -0.5], [ 0.5, -0.5, -0.5]
    ])

    fig = plt.figure(figsize=(6, 6))
    ax = fig.add_subplot(111, projection="3d", computed_zorder=False)

    names = [
        "relative x translation", "relative y translation", "relative z translation",
        "relative x rotation", "relative y rotation", "relative z rotation",
    ]

    def update(frame):
        s = amplitude * np.sin(2*np.pi * frame / nframes)

        RA_rot = expm(s * cross_matrix(thA))
        RB_rot = expm(s * cross_matrix(thB))
        R_mean = expm(s * cross_matrix((thA+thB)/2))
        Rb = _block_diag_rot(R_mean)
        K_t = Rb.T @ K @ Rb

        xA_t = xA + s*uA
        xB_t = xB + s*uB
        VA = xA_t + (RA_rot @ vertices.T).T
        VB = xB_t + (RB_rot @ vertices.T).T

        ax.clear()
        _draw_scene_helix(ax, VA, VB, K_t, xA_t, xB_t, cube_alpha=cube_alpha, **pipe_kwargs)

        ax.scatter(*xA_t, s=40, c='tab:blue', depthshade=False)
        ax.scatter(*xB_t, s=40, c='tab:orange', depthshade=False)

        ax.set_xlim(-1.5, 1.5)
        ax.set_ylim(-1.5, 1.5)
        zmin = min(-1.0, d[2] - 1.0)
        zmax = max( 1.0, d[2] + 1.0)
        ax.set_zlim(zmin, zmax)
        ax.set_xlabel("x"); ax.set_ylabel("y"); ax.set_zlabel("z")
        ax.set_title(names[coordinate])
        ax.set_box_aspect([3, 3, max(3, zmax-zmin)])

    ani = FuncAnimation(fig, update, frames=nframes, interval=interval, blit=False)
    plt.close(fig)
    return HTML(ani.to_jshtml())
```
