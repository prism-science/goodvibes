# Explore bond restraints in Gemmi


```python
import gemmi
```

```python
coordinate_file = 'test_data/7TX0.cif'
st = gemmi.read_structure(coordinate_file)
st.setup_entities()  # supposed to be good practice

monlib_path = '/Applications/ccp4-9/lib/data/monomers'
monlib = gemmi.MonLib()
resnames = st[0].get_all_residue_names()
monlib.read_monomer_lib(monlib_path, resnames)
monlib.update_old_atom_names(st)

topo = gemmi.prepare_topology(st, monlib, h_change=gemmi.HydrogenChange.ReAddButWater)
```

```python
print(len(topo.angles))
for j in range(5):
    ang = topo.angles[j]
    restr = ang.restr
    print(restr, restr.value, restr.esd)
```

```python
print(len(topo.bonds))
for j in range(5):
    bond = topo.bonds[j]
    restr = bond.restr
    print(restr, restr.value, restr.esd)
```

```python
print(len(topo.planes))
for j in range(5):
    plane = topo.planes[j]
    restr = plane.restr
    print(restr.label, restr, '?', restr.esd)
```

```python
print(len(topo.torsions))
for j in range(5):
    torsion = topo.torsions[j]
    restr = torsion.restr
    print(restr.label, restr, restr.value, restr.esd, restr.period)
```

```python
for j in range(5):
    chir = topo.chirs[j]
    restr = chir.restr
    print(restr, restr.sign)
```
